```python
import re

def transform_sender_to_category(sender):
    # 1. Standardize the input
    s = str(sender).strip().lower()

    # 2. Clean numeric strings INCLUDING asterisks for length checks
    # This ensures masked numbers like +639*****98 stay 10+ characters
    masked_and_digits = re.sub(r'[^0-9*]', '', s)

    # --- STEP 1: Handle Privacy Placeholders & Specific Keywords ---
    if "redacted_individual" in s or "redacted_business" in s:
        return "Personal_Mobile"

    if s in ["unknown", "hidden", "private"]:
        return "Anonymized_Sender"

    # --- STEP 2: Handle Numbers (Updated to include masked numbers) ---
    # Mobile numbers (including those with +63 or asterisks) will be 10+ chars
    if len(masked_and_digits) >= 10:
        return "Personal_Mobile"

    # 3 to 6 digit numbers (Clean Short Codes)
    if masked_and_digits and len(masked_and_digits) < 10:
        return "Short_Code"

    # --- STEP 3: Handle Verified Brands ---
    if any(char.isalpha() for char in s):
        return "Verified_Brand"

    return "Unknown"

# --- TEST LOGIC ---
test_data = ["09171234567", "+639*****98", "redacted_individual", "BDO", "8080"]
for item in test_data:
    print(f"{item} -> {transform_sender_to_category(item)}")
```

```python
import pandas as pd

def transform_to_binary_target(df):
    """
    Transforms multi-class categories into binary labels:
    1 = Smishing (Spam)
    0 = Safe (Ads, Gov, OTP, Personal)
    """
    # Define the mapping
    # 'spam' becomes 1, everything else becomes 0
    mapping = {
        'spam': 1,
        'ads': 0,
        'gov': 0,
        'otp': 0,
        'personal': 0
    }

    # Create the new label column
    # .get() handles any unexpected categories by defaulting them to 0 (Safe)
    df['label'] = df['category'].str.lower().str.strip().map(mapping).fillna(0).astype(int)

    return df

# Apply to your filtered dataset
df_filtered = transform_to_binary_target(df_filtered)

# Verify the transformation
print(df_filtered[['category', 'label']].value_counts())
```

```python
import pandas as pd
import re

def extract_smishing_features(df):
    # 1. Text Length
    df['text_len'] = df['text'].apply(len)

    # 2. Contains URL (Binary: 1 if it has http/https/bit.ly/etc)
    url_pattern = r'http[s]?://(?:[a-zA-Z]|[0-9]|[$-_@.&+]|[!*\(\),]|(?:%[0-9a-fA-F][0-9a-fA-F]))+'
    df['has_url'] = df['text'].apply(lambda x: 1 if re.search(url_pattern, x) else 0)

    # 3. Urgency Score (Count of specific words)
    urgency_words = ['voucher', 'congratulations', 'locked', 'winner', 'urgent', 'account', 'verify', 'nanalo', 'bonus', 'com', 'free', 'win', 'deposit', 'claim', 'araw', '100', 'real', 'tv', 'account', 'libreng', 'receive', 'makakuha']
    df['urgency_score'] = df['text'].apply(lambda x: sum(1 for word in urgency_words if word in x.lower()))

    # 4. Sender Category Encoding (Categorical to Numeric)
    # Using Label Encoding or One-Hot Encoding
    df = pd.get_dummies(df, columns=['sender_category'], prefix='sc')

    # 5. Extract Hour
    df['date-received'] = pd.to_datetime(df['date-received'], format='mixed')
    df['hour_received'] = df['date-received'].dt.hour

    return df

# Apply the function
df_features = extract_smishing_features(df_filtered)
```

```python
# ================================
# FEATURE MATRICES FOR MODELING
# ================================

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split

# Labels
y = df["label"]

# Metadata columns
meta_columns = [
    "text_len",
    "has_url",
    "urgency_score",
    "hour_received"
] + [col for col in df.columns if col.startswith("sc_")]

X_meta = df[meta_columns].values
X_text_raw = df["text"]

# Train-test split first to avoid TF-IDF data leakage
X_text_train_raw, X_text_test_raw, X_meta_train, X_meta_test, y_train, y_test = train_test_split(
    X_text_raw,
    X_meta,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

# Fit TF-IDF only on training text
vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    max_features=10000,
    stop_words='english'
)

X_text_train = vectorizer.fit_transform(X_text_train_raw)
X_text_test = vectorizer.transform(X_text_test_raw)

print("X_text_train:", X_text_train.shape)
print("X_text_test:", X_text_test.shape)
print("X_meta_train:", X_meta_train.shape)
print("X_meta_test:", X_meta_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

# 1. Train a linear model to extract feature importance
# Using the training data generated in your previous block
logreg = LogisticRegression(max_iter=1000)
logreg.fit(X_text_train, y_train)

# 2. Map the feature names (words) to the coefficients
feature_names = vectorizer.get_feature_names_out()
coefficients = logreg.coef_[0]

# 3. Create a DataFrame for easier sorting
feature_importance = pd.DataFrame({
    'word': feature_names,
    'coefficient': coefficients
})

# 4. Sort to find the top 20 for each category
# Positive coefficients indicate Smishing (label 1), negative indicate Safe (label 0)
top_smishing = feature_importance.sort_values(by='coefficient', ascending=False).head(20)
top_safe = feature_importance.sort_values(by='coefficient', ascending=True).head(20)

# 5. Display the results
print("--- Top 20 Words Associated with SMISHING ---")
print(top_smishing[['word', 'coefficient']])

print("\n--- Top 20 Words Associated with SAFE ---")
print(top_safe[['word', 'coefficient']])

# Optional: Visualization
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(15, 8))

ax[0].barh(top_smishing['word'], top_smishing['coefficient'], color='red')
ax[0].set_title('Top 20 Smishing Indicators')
ax[0].invert_yaxis()

ax[1].barh(top_safe['word'], top_safe['coefficient'], color='green')
ax[1].set_title('Top 20 Safe Message Indicators')
ax[1].invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

models = {
    "SVM (Text)": (
        SVC(kernel="linear", random_state=42),
        X_text_train,
        X_text_test
    ),

    "Naive Bayes (Text)": (
        MultinomialNB(),
        X_text_train,
        X_text_test
    ),

    "Logistic Regression (Text)": (
        LogisticRegression(max_iter=1000, random_state=42),
        X_text_train,
        X_text_test
    ),

    "Random Forest (Metadata)": (
        RandomForestClassifier(
            n_estimators=100,
            random_state=42
        ),
        X_meta_train,
        X_meta_test
    ),

    "XGBoost (Metadata Base)": (
        XGBClassifier(
            objective="binary:logistic",
            eval_metric="logloss",
            random_state=42
        ),
        X_meta_train,
        X_meta_test
    ),
}

results = []

plt.figure(figsize=(18, 12))

print("--- Individual Model Performance ---")

for i, (name, (model, train_data, test_data)) in enumerate(models.items(), 1):

    model.fit(train_data, y_train)
    preds = model.predict(test_data)

    # Metrics
    acc = accuracy_score(y_test, preds)
    precision = precision_score(y_test, preds)
    recall = recall_score(y_test, preds)
    f1 = f1_score(y_test, preds)

    cm = confusion_matrix(y_test, preds)

    # Store results
    results.append({
        "Model": name,
        "Accuracy": acc,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })

    # Confusion Matrix Plot
    plt.subplot(2, 3, i)

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        cbar=False
    )

    plt.title(
        f"{name}\n"
        f"Acc={acc:.3f} | Prec={precision:.3f}\n"
        f"Recall={recall:.3f} | F1={f1:.3f}"
    )

    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    print(f"{name}: Done")

plt.tight_layout()
plt.show()

# Results table
results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="F1-Score",
    ascending=False
).reset_index(drop=True)

print("\n--- Model Performance Summary ---")
print(results_df)
```

```python
# ==========================================
# STACKED ENSEMBLE APPROACH
# Evaluates all possible non-empty combinations
# 5 base models = 31 stacked ensemble combinations
# ==========================================

import numpy as np
import pandas as pd
import itertools

from sklearn.model_selection import cross_val_predict
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

# -------------------------------
# 1. Define all base models
# -------------------------------

base_models = {
    "SVM": (
        SVC(kernel="linear", probability=True, random_state=42),
        X_text_train,
        X_text_test
    ),

    "LR": (
        LogisticRegression(max_iter=1000, random_state=42),
        X_text_train,
        X_text_test
    ),

    "NB": (
        MultinomialNB(),
        X_text_train,
        X_text_test
    ),

    "RF": (
        RandomForestClassifier(n_estimators=100, random_state=42),
        X_meta_train,
        X_meta_test
    ),

    "XGB_Base": (
        XGBClassifier(
            objective="binary:logistic",
            eval_metric="logloss",
            random_state=42
        ),
        X_meta_train,
        X_meta_test
    )
}

# -------------------------------
# 2. Generate base model probabilities
# -------------------------------

train_probs = {}
test_probs = {}

print("Generating base model probabilities...")

for name, (model, train_data, test_data) in base_models.items():

    train_probs[name] = cross_val_predict(
        model,
        train_data,
        y_train,
        cv=5,
        method="predict_proba"
    )[:, 1]

    model.fit(train_data, y_train)
    test_probs[name] = model.predict_proba(test_data)[:, 1]

    print(f"{name}: probabilities generated")

# -------------------------------
# 3. Evaluate all 31 combinations
# -------------------------------

ensemble_results = []

model_names = list(base_models.keys())

print("\nEvaluating all stacked ensemble combinations...")

for r in range(1, len(model_names) + 1):

    for combo in itertools.combinations(model_names, r):

        combo_name = " + ".join(combo)

        # Stack predicted probabilities as meta-features
        meta_X_train = np.column_stack([
            train_probs[m] for m in combo
        ])

        meta_X_test = np.column_stack([
            test_probs[m] for m in combo
        ])

        # XGBoost meta-classifier
        meta_clf = XGBClassifier(
            objective="binary:logistic",
            eval_metric="logloss",
            random_state=42
        )

        meta_clf.fit(meta_X_train, y_train)

        y_pred = meta_clf.predict(meta_X_test)

        # Evaluation metrics
        acc = accuracy_score(y_test, y_pred)
        precision = precision_score(y_test, y_pred)
        recall = recall_score(y_test, y_pred)
        f1 = f1_score(y_test, y_pred)

        ensemble_results.append({
            "Combination": combo_name,
            "Model_Count": len(combo),
            "Accuracy": acc,
            "Precision": precision,
            "Recall": recall,
            "F1-Score": f1
        })

# -------------------------------
# 4. Display ranked results
# -------------------------------

stacked_results_df = pd.DataFrame(ensemble_results)

stacked_results_df = stacked_results_df.sort_values(
    by="F1-Score",
    ascending=False
).reset_index(drop=True)

print("\n--- All Stacked Ensemble Results ---")
print(stacked_results_df)

print("\n--- Top 10 Stacked Ensemble Combinations ---")
print(stacked_results_df.head(10))

print(f"\nTotal combinations evaluated: {len(stacked_results_df)}")
```

```python
import math
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
from xgboost import XGBClassifier

n_combos = len(stacked_results_df)
cols = 4
rows = math.ceil(n_combos / cols)

fig, axes = plt.subplots(rows, cols, figsize=(20, rows * 4))
axes = axes.flatten()

print(f"Generating all {n_combos} hybrid ensemble confusion matrices...")

for i, (index, row) in enumerate(stacked_results_df.iterrows()):
    combo_name = row["Combination"]
    combo_models = combo_name.split(" + ")

    meta_X_train_sub = np.column_stack([train_probs[m] for m in combo_models])
    meta_X_test_sub = np.column_stack([test_probs[m] for m in combo_models])

    temp_meta_clf = XGBClassifier(
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42
    )

    temp_meta_clf.fit(meta_X_train_sub, y_train)

    y_pred_combo = temp_meta_clf.predict(meta_X_test_sub)
    cm = confusion_matrix(y_test, y_pred_combo)

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Purples",
        cbar=False,
        ax=axes[i]
    )

    axes[i].set_title(
        f"Rank {i+1}: {combo_name}\n"
        f"Acc: {row['Accuracy']:.4f} | F1: {row['F1-Score']:.4f}",
        fontsize=9
    )

    axes[i].set_xlabel("Predicted")
    axes[i].set_ylabel("Actual")

for j in range(n_combos, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
# ================================
# NEW DATA SIMULATION
# Best Stacked Ensemble:
# SVM + LR + NB + RF + XGB_Base → XGBoost Meta-Classifier
# ================================

import numpy as np
import pandas as pd
import re
from xgboost import XGBClassifier

best_combo = ["SVM", "LR", "NB", "RF", "XGB_Base"]

# Use cross-validated training probabilities from the stacking section
meta_X_train_best = np.column_stack([
    train_probs[m] for m in best_combo
])

best_meta_model = XGBClassifier(
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42
)

best_meta_model.fit(meta_X_train_best, y_train)

# Use the already fitted base models from the 31-combination stacking section
svm_model = base_models["SVM"][0]
lr_model = base_models["LR"][0]
nb_model = base_models["NB"][0]
rf_model = base_models["RF"][0]
xgb_base_model = base_models["XGB_Base"][0]


def classify_new_sms_stacked(message, sender, date_received):
    new_df = pd.DataFrame({
        "text": [message],
        "sender": [sender],
        "date-received": [date_received]
    })

    # Sender category
    new_df["sender_category"] = new_df["sender"].apply(transform_sender_to_category)

    # Metadata features
    new_df["text_len"] = new_df["text"].apply(len)

    url_pattern = r"(http[s]?://\S+|www\.\S+|bit\.ly/\S+|tinyurl\.com/\S+|t\.co/\S+)"

    new_df["has_url"] = new_df["text"].apply(
        lambda x: 1 if re.search(url_pattern, str(x).lower()) else 0
    )

    urgency_words = [
        "voucher",
        "congratulations",
        "locked",
        "winner",
        "urgent",
        "account",
        "verify",
        "nanalo"
    ]

    new_df["urgency_score"] = new_df["text"].apply(
        lambda x: sum(1 for word in urgency_words if word in str(x).lower())
    )

    new_df["date-received"] = pd.to_datetime(
        new_df["date-received"],
        errors="coerce"
    )

    new_df["hour_received"] = (
        new_df["date-received"]
        .dt.hour
        .fillna(0)
        .astype(int)
    )

    # One-hot encode sender category
    new_df = pd.get_dummies(
        new_df,
        columns=["sender_category"],
        prefix="sc"
    )

    # Align metadata columns with training metadata columns
    for col in meta_columns:
        if col not in new_df.columns:
            new_df[col] = 0

    new_X_meta = new_df[meta_columns].values

    # Text features
    new_X_text = vectorizer.transform(new_df["text"])

    # Base model probabilities
    svm_prob = svm_model.predict_proba(new_X_text)[:, 1]
    lr_prob = lr_model.predict_proba(new_X_text)[:, 1]
    nb_prob = nb_model.predict_proba(new_X_text)[:, 1]
    rf_prob = rf_model.predict_proba(new_X_meta)[:, 1]
    xgb_base_prob = xgb_base_model.predict_proba(new_X_meta)[:, 1]

    # Meta-features must follow the same order as best_combo
    new_meta_X = np.column_stack([
        svm_prob,
        lr_prob,
        nb_prob,
        rf_prob,
        xgb_base_prob
    ])

    prediction = best_meta_model.predict(new_meta_X)[0]
    probability = best_meta_model.predict_proba(new_meta_X)[0][1]

    label = "SMISHING" if prediction == 1 else "SAFE"

    sender_category_col = new_df.filter(like="sc_").idxmax(axis=1).iloc[0]
    sender_category = sender_category_col.replace("sc_", "")

    return label, probability, {
        "Sender Category": sender_category,
        "Text Length": int(new_df["text_len"].iloc[0]),
        "Has URL": int(new_df["has_url"].iloc[0]),
        "Urgency Score": int(new_df["urgency_score"].iloc[0]),
        "Hour Received": int(new_df["hour_received"].iloc[0]),
        "SVM Smishing Probability": round(float(svm_prob[0]), 4),
        "LR Smishing Probability": round(float(lr_prob[0]), 4),
        "NB Smishing Probability": round(float(nb_prob[0]), 4),
        "RF Smishing Probability": round(float(rf_prob[0]), 4),
        "XGB_Base Smishing Probability": round(float(xgb_base_prob[0]), 4)
    }


while True:
    print("\n========== STACKED SMS CLASSIFIER ==========")

    user_message = input("Enter SMS message: ")
    user_sender = input("Enter sender: ")
    user_date = input("Enter date and time (YYYY-MM-DD HH:MM:SS): ")

    label, probability, extracted_features = classify_new_sms_stacked(
        user_message,
        user_sender,
        user_date
    )

    print("\n--- Extracted Features and Base Model Outputs ---")
    for key, value in extracted_features.items():
        print(f"{key}: {value}")

    print("\n--- Final Prediction Result ---")
    print("Prediction:", label)
    print(f"Final Smishing Probability: {probability:.4f}")

    again = input("\nClassify another SMS? (y/n): ")

    if again.lower() != "y":
        print("Exiting SMS classifier.")
        break
```
